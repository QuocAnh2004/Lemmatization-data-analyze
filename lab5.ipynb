```python
!pip install -q pypdf
from pypdf import PdfReader
import io

pages = ''
dictionary = {} # Tránh dùng từ khóa trùng tên 'dict'
path = 'thebook.pdf'

reader = PdfReader(path)
page = reader.pages[10]
text = page.extract_text()

for index, character in enumerate(text, 1):
    pages += str(character)
    dictionary[index] = character

result = pages
print(result)
```

```python
import nltk
nltk.download('all')
```

```python
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords

tokens_by_sentence = ''
tokens_by_sentence = sent_tokenize(result)
tokens_by_sentence = [w.replace('\n',' ').replace('- ','')for w in tokens_by_sentence]
print(tokens_by_sentence)
```

```python
from nltk.tokenize import word_tokenize
tokens_by_word = word_tokenize(result)
tokens_by_word = [w.replace('\n', '').replace('- ','') for w in tokens_by_word]
print(tokens_by_word)
```

```python
print(stopwords.words('english'))
```

```python
print(stopwords.words('portuguese')[:10])
```

```python
def remove_stopwords(text):
  stopWords = stopwords.words('english')
  not_stopwords = [word for word in text if not word in stopWords]
  return not_stopwords

result = remove_stopwords(tokens_by_word)
print(result)
```

```python
from nltk.stem.snowball import SnowballStemmer
englishStemmer = SnowballStemmer('english')
words = [word for word in result if word.isalpha()]
stemmed = [englishStemmer.stem(word) for word in tokens_by_word]
print(stemmed[:50])
```

```python
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
lemmatized = [wordnet_lemmatizer.lemmatize(word) for word in tokens_by_word]
print(lemmatized[:50])
```
